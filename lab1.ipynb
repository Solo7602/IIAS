```python
import pandas as pd
from datetime import datetime

# Загрузка данных
data = pd.read_csv('10k_dataset_processed.csv')

def calculate_age(birth_date, comment_date):
    try:
        birth_date = pd.to_datetime(birth_date, errors='coerce')
        comment_date = pd.to_datetime(comment_date, errors='coerce')

        if pd.isna(birth_date) or pd.isna(comment_date):
            return None

        age = comment_date.year - birth_date.year
        # Проверяем, был ли уже день рождения в этом году
        if (comment_date.month < birth_date.month) or \
           (comment_date.month == birth_date.month and comment_date.day < birth_date.day):
            age -= 1
        if age <= 5 or age > 100:
            return None
        return age
    except:
        return None

# 4. Создаем столбец с возрастом
data['Age'] = data.apply(lambda row: calculate_age(row['BirthDay'], row['PostDate']), axis=1)

# Анализ данных перед очисткой
print("\nСтатистика перед обработкой:")
print(f"Всего записей: {len(data)}")
print(f"Успешно вычислено возрастов: {data['Age'].notna().sum()}")
print(f"Невалидных дат: {data['Age'].isna().sum()}")

# Очистка данных
print("\nОчистка данных...")
data = data.drop_duplicates()
data = data[data['Age'].notna()]
print(f"Записей после очистки: {len(data)}")


# Сохранение результатов
output_columns = [col for col in data.columns if col not in ['Age_normalized', 'Age_standardized'] or col in ['Age', 'BirthDay', 'PostDate']]
data.to_csv('10k_dataset_processed_final.csv', index=False, encoding='utf-8')

# Вывод результатов
# print("\nРезультаты обработки:")
# print(data[['BirthDay', 'PostDate', 'Age', 'Age_normalized', 'Age_standardized']].head(10))

print("\nРаспределение возрастов:")
print(data['Age'].value_counts().sort_index())

print("\nОбработка завершена.")
```
