```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import pandas as pd
import numpy as np

# Загрузка данных
data = pd.read_csv('sample_data/10k_dataset_processed_final.csv')
print(data['Age'].value_counts())

# Разбиение на обучающую (70%), тестовую (30%) выборки
train_data, test_data = train_test_split(data, test_size=0.3, random_state=42)

# One-hot Encoding для Sex
encoder = OneHotEncoder(sparse_output=False, drop="first")

encoded_values = encoder.fit_transform(data[["Sex"]])

encoded_columns = encoder.get_feature_names_out(["Sex"])

encoded_values_df = pd.DataFrame(encoded_values, columns=encoded_columns)
data = pd.concat([data, encoded_values_df], axis=1)

# Дискретизация Age
data['Age_Group'] = pd.cut(data['Age'], bins=[0, 15, 40, 80, 100], labels=['[0, 15]', '[15, 40]', '[40, 80]', '[80, 100]'])

interval = {'interval': ['[0, 15]', '[15, 40]', '[40, 80]', '[80, 100]']}
interval_to_category = {
    '[0-15]': 1,
    '[15, 40]': 2,
    '[40, 80]': 4,
    '[80, 100]': 5
}

data['Age_Category'] = data['Age_Group'].map(interval_to_category)

# Ручной синтез
data['Comment_Length'] = data['Comment'].apply(len)
# Масштабирование числовых признаков
scaler = StandardScaler()
data[['Age_Normalize', 'Comment_Length_Normalize']] = scaler.fit_transform(data[['Age', 'Comment_Length']])

print(data.head(20))

data.to_csv('dataset_processed_final_with_ageGroup.csv', index=False, encoding='utf-8')
```
